# Logistic Regression / Linear Classifier trên BOW

```python
import os

# List the contents of the 'data' directory to debug file paths
!ls -R data
```

```python
from datasets import load_dataset

# Load the PIQA dataset using the datasets library
# The dataset was already downloaded and prepared by the previous script execution.
dataset = load_dataset('piqa')

# Access the train, validation (dev), and test splits
train_ds = dataset['train']
# For evaluation purposes, we will use the 'validation' split as the 'test' set,
# since the official 'test' split of PIQA on Hugging Face is unlabelled.
test_ds = dataset['validation']
dev_ds = dataset['validation'] # Keeping dev_ds as validation for consistency if other parts of the code use it for dev.

print("PIQA Train dataset head:")
print(train_ds.to_pandas().head())
```

```python
import pandas as pd

def prepare_data_for_binary_classification(dataset_split):
    combined_texts = []
    binary_labels = []

    for entry in dataset_split:
        goal = entry['goal']
        sol1 = entry['sol1']
        sol2 = entry['sol2']
        label = entry['label']

        # For sol1
        combined_texts.append(f"{goal} {sol1}")
        binary_labels.append(1 if label == 0 else 0) # If sol1 is correct (original label 0), binary label is 1

        # For sol2
        combined_texts.append(f"{goal} {sol2}")
        binary_labels.append(1 if label == 1 else 0) # If sol2 is correct (original label 1), binary label is 1

    return pd.DataFrame({'text': combined_texts, 'label': binary_labels})

# Prepare the training, development, and test data
train_df = prepare_data_for_binary_classification(train_ds)
dev_df = prepare_data_for_binary_classification(dev_ds)
test_df = prepare_data_for_binary_classification(test_ds)

print("Prepared Training Data head:")
print(train_df.head())
print("\nPrepared Training Data Tail:")
print(train_df.tail())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Fit the vectorizer on the training data text and transform it
X_train = vectorizer.fit_transform(train_df['text'])
y_train = train_df['label']

# Transform the development and test data text
X_dev = vectorizer.transform(dev_df['text'])
y_dev = dev_df['label']
X_test = vectorizer.transform(test_df['text'])
y_test = test_df['label']

print("Shape of X_train:", X_train.shape)
print("Shape of X_dev:", X_dev.shape)
print("Shape of X_test:", X_test.shape)
print("Number of features (vocabulary size):", len(vectorizer.vocabulary_))
```

```python
from sklearn.linear_model import LogisticRegression

# Initialize Logistic Regression model
# Set max_iter to a higher value to ensure convergence
model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model on the training data
model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score

# Predict on the test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy on the test set: {accuracy:.4f}")
```

```python
import numpy as np

def get_model_choice(pred_sol1, pred_sol2):
    if pred_sol1 == 1 and pred_sol2 == 0:
        return 'A'
    elif pred_sol1 == 0 and pred_sol2 == 1:
        return 'B'
    elif pred_sol1 == 1 and pred_sol2 == 1:
        return 'Both A and B (Ambiguous)'
    else:
        return 'Neither A nor B (Ambiguous)'

# Generate new random indices based on the current test_ds size (1838 entries)
# The 'replace=False' ensures unique indices are chosen.
random_original_indices = np.random.choice(len(test_ds), size=5, replace=False)

print("Sample Predictions:")
print("=" * 20)

for i, original_idx in enumerate(random_original_indices):
    # Cast original_idx to a Python int
    original_entry = test_ds[int(original_idx)]
    question = original_entry['goal']
    sol1 = original_entry['sol1']
    sol2 = original_entry['sol2']
    correct_label = original_entry['label']

    # Map the original_idx to the corresponding indices in test_df and y_pred
    # Each original entry creates two entries in test_df, so we need to multiply by 2
    df_idx_sol1 = int(original_idx * 2)
    df_idx_sol2 = int(original_idx * 2 + 1)

    # Ensure these indices are within the bounds of y_pred before accessing
    if df_idx_sol1 < len(y_pred) and df_idx_sol2 < len(y_pred):
        pred_sol1 = y_pred[df_idx_sol1]
        pred_sol2 = y_pred[df_idx_sol2]

        model_choice = get_model_choice(pred_sol1, pred_sol2)

        print(f"Question {i+1}:")
        print(f"  Q: {question}")
        print(f"  A: {sol1}")
        print(f"  B: {sol2}")
        print(f"  Correct Answer: {('A' if correct_label == 0 else 'B')}")
        print(f"  Model's prediction for A: {pred_sol1}")
        print(f"  Model's prediction for B: {pred_sol2}")
        print(f"  Model chose: {model_choice}")
        print("-" * 20)
    else:
        print(f"Skipping question {i+1} as derived indices ({df_idx_sol1}, {df_idx_sol2}) are out of bounds for y_pred (length {len(y_pred)}).")
        print("-" * 20)
```

# RoBERTa

```python
# Install libraries
!pip install -q transformers datasets scikit-learn

import os
os.environ["TOKENIZERS_PARALLELISM"] = "false"

import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup
)
from datasets import load_dataset
from sklearn.metrics import accuracy_score
from torch.optim import AdamW
from tqdm import tqdm
import numpy as np
import random
from collections import Counter

# ============== Configs ==============
MODEL_NAME = "roberta-base"          # try "distilroberta-base" if training is unstable
BATCH_SIZE = 16
MAX_LEN = 128
EPOCHS = 5
LR = 2e-5
WARMUP_RATIO = 0.1
SEED = 42
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ============== Seed ==============
def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
set_seed(SEED)

# ============== Load dataset ==============
piqa = load_dataset("piqa")
train_raw = piqa["train"]
val_raw = piqa["validation"]

# ============== Tokenizer & Model ==============
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
model.to(DEVICE)

# ============== Build two-sample dataset ==============
def build_binary_dataset(raw_ds):
    samples = []
    for item in raw_ds:
        goal = item["goal"]
        sol1 = item["sol1"]
        sol2 = item["sol2"]
        label = item["label"]  # 0 if sol1 correct, 1 if sol2 correct

        samples.append({
            "text": f"Question: {goal}\nOption: {sol1}",
            "label": 1 if label == 0 else 0
        })
        samples.append({
            "text": f"Question: {goal}\nOption: {sol2}",
            "label": 1 if label == 1 else 0
        })
    return samples

train_data = build_binary_dataset(train_raw)
val_data = build_binary_dataset(val_raw)

# Sanity check: labels should be balanced
print("Train label distribution:", Counter([s["label"] for s in train_data]))
print("Val label distribution:", Counter([s["label"] for s in val_data]))

# ============== Dataset class ==============
class BinaryPIQADataset(Dataset):
    def __init__(self, data, tokenizer, max_len=MAX_LEN):
        self.data = data
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        item = self.data[idx]
        enc = self.tokenizer(
            item["text"],
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt"
        )
        return {
            "input_ids": enc["input_ids"].squeeze(),
            "attention_mask": enc["attention_mask"].squeeze(),
            "labels": torch.tensor(item["label"], dtype=torch.long)
        }

train_ds = BinaryPIQADataset(train_data, tokenizer)
val_ds = BinaryPIQADataset(val_data, tokenizer)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE)

# ============== Optimizer & Scheduler ==============
optimizer = AdamW(model.parameters(), lr=LR)
total_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(WARMUP_RATIO * total_steps),
    num_training_steps=total_steps
)

# ============== Training with early stopping ==============
best_val_acc = 0.0
patience = 2
patience_counter = 0

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS} Training"):
        optimizer.zero_grad()
        input_ids = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        labels = batch["labels"].to(DEVICE)

        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_loss += loss.item()

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch+1} avg train loss: {avg_loss:.4f}")

    # Validation on binary samples (quick sanity)
    model.eval()
    preds, trues = [], []
    with torch.no_grad():
        for batch in tqdm(val_loader, desc="Validation (binary samples)"):
            input_ids = batch["input_ids"].to(DEVICE)
            attention_mask = batch["attention_mask"].to(DEVICE)
            labels = batch["labels"].to(DEVICE)

            logits = model(input_ids, attention_mask=attention_mask).logits
            pred = torch.argmax(logits, dim=1)
            preds.extend(pred.cpu().numpy())
            trues.extend(labels.cpu().numpy())

    val_acc_binary = accuracy_score(trues, preds)
    print(f"Val accuracy (binary samples): {val_acc_binary:.4f}")

    # True PIQA evaluation: pairwise scoring per question
    correct, total = 0, 0
    with torch.no_grad():
        for item in val_raw:
            goal, sol1, sol2, label = item["goal"], item["sol1"], item["sol2"], item["label"]

            inp1 = tokenizer(f"Question: {goal}\nOption: {sol1}",
                             truncation=True, padding="max_length", max_length=MAX_LEN,
                             return_tensors="pt").to(DEVICE)
            inp2 = tokenizer(f"Question: {goal}\nOption: {sol2}",
                             truncation=True, padding="max_length", max_length=MAX_LEN,
                             return_tensors="pt").to(DEVICE)

            out1 = model(**inp1).logits
            out2 = model(**inp2).logits

            score1 = torch.softmax(out1, dim=1)[0][1].item()  # prob "correct"
            score2 = torch.softmax(out2, dim=1)[0][1].item()

            pred = 0 if score1 > score2 else 1
            if pred == label:
                correct += 1
            total += 1

    val_acc_pairwise = correct / total
    print(f"Val accuracy (pairwise PIQA): {val_acc_pairwise:.4f}")

    # Early stopping on pairwise accuracy
    if val_acc_pairwise > best_val_acc:
        best_val_acc = val_acc_pairwise
        patience_counter = 0
        torch.save(model.state_dict(), "best_roberta_piqa.bin")
        print(f"✅ New best pairwise acc: {best_val_acc:.4f} — model saved.")
    else:
        patience_counter += 1
        print(f"⏳ No improvement. Patience: {patience_counter}/{patience}")
        if patience_counter >= patience:
            print("🛑 Early stopping triggered.")
            break

# ============== Load best model ==============
model.load_state_dict(torch.load("best_roberta_piqa.bin", map_location=DEVICE))
model.eval()

# ============== Print 5 sample predictions ==============
print("\nSample Predictions (5 random questions):")
indices = np.random.choice(len(val_raw), size=5, replace=False)

for i, idx in enumerate(indices):
    item = val_raw[int(idx)]
    goal, sol1, sol2, label = item["goal"], item["sol1"], item["sol2"], item["label"]

    inp1 = tokenizer(f"Question: {goal}\nOption: {sol1}",
                     truncation=True, padding="max_length", max_length=MAX_LEN,
                     return_tensors="pt").to(DEVICE)
    inp2 = tokenizer(f"Question: {goal}\nOption: {sol2}",
                     truncation=True, padding="max_length", max_length=MAX_LEN,
                     return_tensors="pt").to(DEVICE)

    with torch.no_grad():
        out1 = model(**inp1).logits
        out2 = model(**inp2).logits

    score1 = torch.softmax(out1, dim=1)[0][1].item()
    score2 = torch.softmax(out2, dim=1)[0][1].item()

    pred_choice = "A" if score1 > score2 else "B"
    correct_choice = "A" if label == 0 else "B"

    print("="*50)
    print(f"Question {i+1}: {goal}")
    print(f"Option A: {sol1}")
    print(f"Option B: {sol2}")
    print(f"Model Prediction: {pred_choice} (scoreA={score1:.3f}, scoreB={score2:.3f})")
    print(f"Correct Answer: {correct_choice}")
```

# DistilBERT

```python
from transformers import AutoTokenizer

# Initialize a pre-trained DistilBERT tokenizer
tokenizer_distilbert = AutoTokenizer.from_pretrained('distilbert-base-uncased')

print("DistilBERT tokenizer loaded successfully.")
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader

def tokenize_function(texts):
    return tokenizer_distilbert(texts, padding='max_length', truncation=True, return_tensors='pt', max_length=128)

# Tokenize datasets
print("Tokenizing training data with DistilBERT tokenizer...")
train_encodings_distilbert = tokenize_function(train_df['text'].tolist())
print("Tokenizing development data with DistilBERT tokenizer...")
dev_encodings_distilbert = tokenize_function(dev_df['text'].tolist())
print("Tokenizing test data with DistilBERT tokenizer...")
test_encodings_distilbert = tokenize_function(test_df['text'].tolist())

# Convert labels to PyTorch tensors
train_labels_distilbert = torch.tensor(train_df['label'].tolist())
dev_labels_distilbert = torch.tensor(dev_df['label'].tolist())
test_labels_distilbert = torch.tensor(test_df['label'].tolist())

# Create PyTorch TensorDataset
train_dataset_distilbert = TensorDataset(train_encodings_distilbert['input_ids'], train_encodings_distilbert['attention_mask'], train_labels_distilbert)
dev_dataset_distilbert = TensorDataset(dev_encodings_distilbert['input_ids'], dev_encodings_distilbert['attention_mask'], dev_labels_distilbert)
test_dataset_distilbert = TensorDataset(test_encodings_distilbert['input_ids'], test_encodings_distilbert['attention_mask'], test_labels_distilbert)

# Create DataLoaders
batch_size = 32
train_loader_distilbert = DataLoader(train_dataset_distilbert, batch_size=batch_size, shuffle=True)
dev_loader_distilbert = DataLoader(dev_dataset_distilbert, batch_size=batch_size, shuffle=False)
test_loader_distilbert = DataLoader(test_dataset_distilbert, batch_size=batch_size, shuffle=False)

print("Data tokenized and DataLoaders created successfully for DistilBERT.")
print(f"Number of training batches (DistilBERT): {len(train_loader_distilbert)}")
print(f"Number of development batches (DistilBERT): {len(dev_loader_distilbert)}")
print(f"Number of test batches (DistilBERT): {len(test_loader_distilbert)}")
```

```python
from transformers import DistilBertForSequenceClassification

# Load a pre-trained DistilBERT model for sequence classification
# num_labels=2 for binary classification (correct/incorrect answer)
model_distilbert = DistilBertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)

print("DistilBERT model for sequence classification loaded successfully.")
```

```python
import torch
from torch.optim import AdamW
from transformers import get_linear_schedule_with_warmup
from tqdm.notebook import tqdm

# Set device to GPU if available, else CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model_distilbert.to(device)

# Training parameters
epochs = 3
learning_rate = 2e-5

# Optimizer and learning rate scheduler
optimizer_distilbert = AdamW(model_distilbert.parameters(), lr=learning_rate)
total_steps_distilbert = len(train_loader_distilbert) * epochs
scheduler_distilbert = get_linear_schedule_with_warmup(optimizer_distilbert, num_warmup_steps=0, num_training_steps=total_steps_distilbert)

# Training loop
model_distilbert.train()
print("Starting DistilBERT model training...")
for epoch in range(epochs):
    print(f"\nEpoch {epoch + 1}/{epochs}")
    total_loss_distilbert = 0
    for batch in tqdm(train_loader_distilbert, desc="Training DistilBERT"):
        optimizer_distilbert.zero_grad()

        input_ids = batch[0].to(device)
        attention_mask = batch[1].to(device)
        labels = batch[2].to(device)

        outputs = model_distilbert(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_loss_distilbert += loss.item()

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model_distilbert.parameters(), 1.0) # Clip gradients to prevent exploding gradients
        optimizer_distilbert.step()
        scheduler_distilbert.step()

    avg_train_loss_distilbert = total_loss_distilbert / len(train_loader_distilbert)
    print(f"  Average training loss (DistilBERT): {avg_train_loss_distilbert:.4f}")

print("DistilBERT model training complete.")
```
